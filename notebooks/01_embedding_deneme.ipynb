```python
from sentence_transformers import SentenceTransformer

model = SentenceTransformer('paraphrase-multilingual-MiniLM-L12-v2')

cumleler = [
    "Kedi bahçede uyuyor.",
    "Bahçede bir kedi uyumakta.",
    "Otomobil motoru arızalandı."
]

vektorler = model.encode(cumleler)
print(vektorler.shape)
```

```python
print(model.similarity(vektorler, vektorler))
```

```python
import numpy as np

ic_carpim = np.dot(vektorler[0], vektorler[1])
uzunluk_a = np.linalg.norm(vektorler[0])
uzunluk_b = np.linalg.norm(vektorler[1])

benzerlik = ic_carpim / (uzunluk_a * uzunluk_b)

print(benzerlik)
```

```python
bilgiler = [
    "Ahmed, Gazi Üniversitesi'nde Elektrik-Elektronik Mühendisliği 3. sınıf öğrencisidir.",
    "Ahmed üniversiteye diş hekimliğinde başladı, 2024'te Gazi EEE'ye geçiş yaptı.",
    "Ahmed'in ana ilgi alanı Edge AI ve gömülü yapay zeka; modelleri mikrodenetleyiciler gibi donanımlarda çalıştırmak istiyor.",
    "Ahmed gömülü sistemler üzerine bir mentorla çalışıyor ve C dilinde pointer konularında deneyim kazandı.",
    "Ahmed kariyerini ASELSAN, ROKETSAN ve TUSAŞ gibi savunma sanayii şirketlerinde yapmak istiyor.",
    "Ahmed yüksek lisansını Almanya'da yapmayı düşünüyor.",
    "Ahmed düzenli olarak spor salonuna gidiyor."
]
```

```python
bilgi_vektorleri = model.encode(bilgiler)

soru = "Ahmed hangi üniversitede okuyor?"
soru_vektoru = model.encode([soru])

skorlar = model.similarity(soru_vektoru, bilgi_vektorleri)
print(skorlar)

en_iyi = skorlar[0].argmax()
print(bilgiler[en_iyi])

import torch

degerler, siralar = torch.topk(skorlar[0], k=3)
for skor, i in zip(degerler, siralar):
    print(f"{skor:.3f}  {bilgiler[i]}")
```
